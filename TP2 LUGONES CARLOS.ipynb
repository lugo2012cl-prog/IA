En este ejercicio deben implementar un algoritmo de búsqueda que no sea **Búsqueda Primero en Anchura (BFS)** para resolver el problema de la Torre de Hanoi. La nota máxima dependerá del algoritmo implementado:

- **Búsqueda Primero en Profundidad**: nota máxima 6.
- **Búsqueda de Costo Uniforme**: nota máxima 6.
- **Búsqueda de Profundidad Limitada con Profundidad Iterativa**: nota máxima 7.
- **Búsqueda Voraz usando la heurística dada en el aula virtual**: nota máxima 8.
- **Búsqueda Voraz usando una heurística desarrollada por vos**: nota máxima 9.

- **Búsqueda A\* usando una heurística desarrollada por vos**: nota máxima 10.

La función debe devolver la salida correspondiente a la solución encontrada o `None si no se encontró una solución.

Además, debe calcular métricas de rendimiento que, como mínimo, incluyan:

- `solution_found`: `True` si se encontró la solución, `False` en caso contrario.
- `nodes_explored`: cantidad de nodos explorados (entero).
- `states_visited`: cantidad de estados distintos visitados (entero).
- `nodes_in_frontier`: cantidad de nodos que quedaron en la frontera al finalizar la ejecución (entero).
- `max_depth`: máxima profundidad explorada (entero).
- `cost_total`: costo total para encontrar la solución (float).

```python
from aima_libs.hanoi_states import ProblemHanoi, StatesHanoi
from aima_libs.tree_hanoi import NodeHanoi
```

```python

def search_algorithm(number_disks=5) -> tuple[NodeHanoi, dict]:

    list_disks = [i for i in range(number_disks, 0, -1)]
    initial_state = StatesHanoi(list_disks, [], [], max_disks=number_disks)
    goal_state = StatesHanoi([], [], list_disks, max_disks=number_disks)
    problem = ProblemHanoi(initial=initial_state, goal=goal_state)

    ##### EDITAR ESTA ZONA
    import heapq
    import itertools

    def get_rods(state):

        if hasattr(state, "rods"):
            return state.rods
        if hasattr(state, "state"):
            return state.state
        return (state.rod1, state.rod2, state.rod3)

    def is_goal(state):
        if hasattr(problem, "goal_test"):
            return problem.goal_test(state)
        return problem.is_goal(state)

    def heuristic(state):
        """
        Heurística admisible: recorre los discos de mayor a menor.
        En cuanto encuentra uno mal ubicado, ese disco y todos los más
        chicos quedan marcados como 'a mover', porque inevitablemente
        habrá que sacarlos del medio para destrabar al disco grande.
        """
        rods_s = get_rods(state)
        rods_g = get_rods(goal_state)

        peg_of_state = {}
        for peg_idx, rod in enumerate(rods_s):
            for disk in rod:
                peg_of_state[disk] = peg_idx

        peg_of_goal = {}
        for peg_idx, rod in enumerate(rods_g):
            for disk in rod:
                peg_of_goal[disk] = peg_idx

        misplaced = 0
        contaminado = False
        for disk in range(number_disks, 0, -1):
            if contaminado or peg_of_state.get(disk) != peg_of_goal.get(disk):
                misplaced += 1
                contaminado = True

        return misplaced

    # Inicializamos las salidas, pero reemplazar con lo que se quiera usar.
    metrics = {
        "solution_found": False,
        "nodes_explored": 0,
        "states_visited": 0,
        "nodes_in_frontier": 0,
        "max_depth": 0,
        "cost_total": 0.0,
    }

    root = NodeHanoi(initial_state)

    counter = itertools.count()
    frontier = []
    h0 = heuristic(root.state)
    heapq.heappush(frontier, (root.path_cost + h0, next(counter), root))

    best_g = {root.state: root.path_cost}
    explored = set()

    solution = None
    max_depth = 0

    # TODO: Completar con el algoritmo de búsqueda que desees implementar
    while frontier:
        f, _, node = heapq.heappop(frontier)

        if node.state in explored:
            continue

        metrics["nodes_explored"] += 1
        max_depth = max(max_depth, node.depth)

        if is_goal(node.state):
            solution = node
            metrics["solution_found"] = True
            metrics["cost_total"] = node.path_cost
            break

        explored.add(node.state)

        for child in node.expand(problem):
            if child.state in explored:
                continue

            g = child.path_cost
            if child.state not in best_g or g < best_g[child.state]:
                best_g[child.state] = g
                h = heuristic(child.state)
                heapq.heappush(frontier, (g + h, next(counter), child))

    metrics["states_visited"] = len(explored)
    metrics["nodes_in_frontier"] = len(frontier)
    metrics["max_depth"] = max_depth

    #####

    return solution, metrics
```

```python
# =====================================================================
# COMPARACIÓN DE 7 ALGORITMOS DE BÚSQUEDA - TORRES DE HANOI
# BFS, DFS, UCS, DLS, IDDFS, Greedy (voraz) y A*
# =====================================================================

import heapq
import itertools
import time
from collections import deque
import matplotlib.pyplot as plt
import matplotlib.patches as patches
from matplotlib.animation import FuncAnimation
from IPython.display import HTML

from aima_libs.hanoi_states import ProblemHanoi, StatesHanoi
from aima_libs.tree_hanoi import NodeHanoi


# ---------------------------------------------------------------------
# Helpers comunes
# ---------------------------------------------------------------------

def get_rods(state):
    if hasattr(state, "rods"):
        return state.rods
    if hasattr(state, "state"):
        return state.state
    return (state.rod1, state.rod2, state.rod3)


def is_goal(problem, state):
    if hasattr(problem, "goal_test"):
        return problem.goal_test(state)
    return problem.is_goal(state)


def heuristic(state, goal_state, number_disks):
    """
    Heurística admisible: recorre los discos de mayor a menor.
    En cuanto encuentra uno mal ubicado, ese disco y todos los más
    chicos quedan marcados como 'a mover', porque inevitablemente
    habrá que sacarlos del medio para destrabar al disco grande.
    """
    rods_s, rods_g = get_rods(state), get_rods(goal_state)
    peg_s = {d: p for p, rod in enumerate(rods_s) for d in rod}
    peg_g = {d: p for p, rod in enumerate(rods_g) for d in rod}
    misplaced, contaminado = 0, False
    for disk in range(number_disks, 0, -1):
        if contaminado or peg_s.get(disk) != peg_g.get(disk):
            misplaced += 1
            contaminado = True
    return misplaced


def build_problem(number_disks):
    list_disks = [i for i in range(number_disks, 0, -1)]
    initial_state = StatesHanoi(list_disks, [], [], max_disks=number_disks)
    goal_state = StatesHanoi([], [], list_disks, max_disks=number_disks)
    problem = ProblemHanoi(initial=initial_state, goal=goal_state)
    return problem, initial_state, goal_state


def base_metrics():
    return {
        "solution_found": False,
        "nodes_explored": 0,
        "states_visited": 0,
        "nodes_in_frontier": 0,
        "max_depth": 0,
        "cost_total": 0.0,
    }


# ---------------------------------------------------------------------
# 1) BFS (en anchura) cola FIFO, explora por niveles
# ---------------------------------------------------------------------

def run_bfs(number_disks):
    problem, initial_state, goal_state = build_problem(number_disks)
    root = NodeHanoi(initial_state)
    frontier = deque([root])
    explored = {root.state}
    metrics = base_metrics()
    solution = None
    while frontier:
        node = frontier.popleft()
        metrics["nodes_explored"] += 1
        metrics["max_depth"] = max(metrics["max_depth"], node.depth)
        if is_goal(problem, node.state):
            solution = node
            metrics["solution_found"] = True
            metrics["cost_total"] = node.path_cost
            break
        for child in node.expand(problem):
            if child.state not in explored:
                explored.add(child.state)
                frontier.append(child)
    metrics["states_visited"] = len(explored)
    metrics["nodes_in_frontier"] = len(frontier)
    return solution, metrics


# ---------------------------------------------------------------------
# 2) DFS (en profundidad) pila LIFO
# ---------------------------------------------------------------------

def run_dfs(number_disks):
    problem, initial_state, goal_state = build_problem(number_disks)
    root = NodeHanoi(initial_state)
    frontier = [root]
    explored = set()
    metrics = base_metrics()
    solution = None
    while frontier:
        node = frontier.pop()
        if node.state in explored:
            continue
        explored.add(node.state)
        metrics["nodes_explored"] += 1
        metrics["max_depth"] = max(metrics["max_depth"], node.depth)
        if is_goal(problem, node.state):
            solution = node
            metrics["solution_found"] = True
            metrics["cost_total"] = node.path_cost
            break
        for child in node.expand(problem):
            if child.state not in explored:
                frontier.append(child)
    metrics["states_visited"] = len(explored)
    metrics["nodes_in_frontier"] = len(frontier)
    return solution, metrics


# ---------------------------------------------------------------------
# 3) UCS (costo uniforme) cola de prioridad ordenada solo por g 
# ---------------------------------------------------------------------

def run_ucs(number_disks):
    problem, initial_state, goal_state = build_problem(number_disks)
    root = NodeHanoi(initial_state)
    counter = itertools.count()
    frontier = [(root.path_cost, next(counter), root)]
    best_g = {root.state: root.path_cost}
    explored = set()
    metrics = base_metrics()
    solution = None
    while frontier:
        g, _, node = heapq.heappop(frontier)
        if node.state in explored:
            continue
        metrics["nodes_explored"] += 1
        metrics["max_depth"] = max(metrics["max_depth"], node.depth)
        if is_goal(problem, node.state):
            solution = node
            metrics["solution_found"] = True
            metrics["cost_total"] = node.path_cost
            break
        explored.add(node.state)
        for child in node.expand(problem):
            if child.state in explored:
                continue
            gc = child.path_cost
            if child.state not in best_g or gc < best_g[child.state]:
                best_g[child.state] = gc
                heapq.heappush(frontier, (gc, next(counter), child))
    metrics["states_visited"] = len(explored)
    metrics["nodes_in_frontier"] = len(frontier)
    return solution, metrics


# ---------------------------------------------------------------------
# 4) DLS es DFS con límite de profundidad FIJO (no iterativo)
# ---------------------------------------------------------------------

def run_dls(number_disks, limit=None):
    if limit is None:
        limit = 2 ** number_disks  
    problem, initial_state, goal_state = build_problem(number_disks)
    root = NodeHanoi(initial_state)
    frontier = [root]
    best_depth = {root.state: root.depth}
    metrics = base_metrics()
    solution = None
    while frontier:
        node = frontier.pop()
        metrics["nodes_explored"] += 1
        metrics["max_depth"] = max(metrics["max_depth"], node.depth)
        if is_goal(problem, node.state):
            solution = node
            metrics["solution_found"] = True
            metrics["cost_total"] = node.path_cost
            break
        if node.depth < limit:
            for child in node.expand(problem):
                if child.state not in best_depth or best_depth[child.state] > child.depth:
                    best_depth[child.state] = child.depth
                    frontier.append(child)
    metrics["states_visited"] = len(best_depth)
    metrics["nodes_in_frontier"] = len(frontier)
    return solution, metrics


# ---------------------------------------------------------------------
# 5) IDDFS: DFS con límite creciente/iterativa 
# ---------------------------------------------------------------------

def run_iddfs(number_disks, max_limit=60):
    problem, initial_state, goal_state = build_problem(number_disks)
    metrics = base_metrics()
    solution = None
    for limit in range(0, max_limit + 1):
        root = NodeHanoi(initial_state)
        frontier = [root]
        best_depth = {}
        found = None
        while frontier:
            node = frontier.pop()
            if node.state in best_depth and best_depth[node.state] <= node.depth:
                continue
            best_depth[node.state] = node.depth
            metrics["nodes_explored"] += 1
            metrics["max_depth"] = max(metrics["max_depth"], node.depth)
            if is_goal(problem, node.state):
                found = node
                break
            if node.depth < limit:
                for child in node.expand(problem):
                    if child.state not in best_depth or best_depth[child.state] > child.depth:
                        frontier.append(child)
        metrics["states_visited"] += len(best_depth)
        if found is not None:
            solution = found
            metrics["solution_found"] = True
            metrics["cost_total"] = found.path_cost
            break
    return solution, metrics


# ---------------------------------------------------------------------
# 6) Greedy (voraz): cola de prioridad ordenada solo por h (heurística)
# ---------------------------------------------------------------------

def run_greedy(number_disks):
    problem, initial_state, goal_state = build_problem(number_disks)
    root = NodeHanoi(initial_state)
    counter = itertools.count()
    frontier = [(heuristic(root.state, goal_state, number_disks), next(counter), root)]
    explored = set()
    metrics = base_metrics()
    solution = None
    while frontier:
        h, _, node = heapq.heappop(frontier)
        if node.state in explored:
            continue
        explored.add(node.state)
        metrics["nodes_explored"] += 1
        metrics["max_depth"] = max(metrics["max_depth"], node.depth)
        if is_goal(problem, node.state):
            solution = node
            metrics["solution_found"] = True
            metrics["cost_total"] = node.path_cost
            break
        for child in node.expand(problem):
            if child.state not in explored:
                hc = heuristic(child.state, goal_state, number_disks)
                heapq.heappush(frontier, (hc, next(counter), child))
    metrics["states_visited"] = len(explored)
    metrics["nodes_in_frontier"] = len(frontier)
    return solution, metrics
    
# ---------------------------------------------------------------------
# 7) A*
# ---------------------------------------------------------------------

def run_astar(number_disks=5) -> tuple[NodeHanoi, dict]:
    list_disks = [i for i in range(number_disks, 0, -1)]
    initial_state = StatesHanoi(list_disks, [], [], max_disks=number_disks)
    goal_state = StatesHanoi([], [], list_disks, max_disks=number_disks)
    problem = ProblemHanoi(initial=initial_state, goal=goal_state)

    def _get_rods(state):
        if hasattr(state, "rods"):
            return state.rods
        if hasattr(state, "state"):
            return state.state
        return (state.rod1, state.rod2, state.rod3)

    def _is_goal(state):
        if hasattr(problem, "goal_test"):
            return problem.goal_test(state)
        return problem.is_goal(state)

    def _heuristic(state):
        """
        Heurística admisible: recorre los discos de mayor a menor.
        En cuanto encuentra uno mal ubicado, ese disco y todos los más
        chicos quedan marcados como 'a mover', porque inevitablemente
        habrá que sacarlos del medio para destrabar al disco grande.
        """
        rods_s = _get_rods(state)
        rods_g = _get_rods(goal_state)
        peg_of_state = {}
        for peg_idx, rod in enumerate(rods_s):
            for disk in rod:
                peg_of_state[disk] = peg_idx
        peg_of_goal = {}
        for peg_idx, rod in enumerate(rods_g):
            for disk in rod:
                peg_of_goal[disk] = peg_idx
        misplaced = 0
        contaminado = False
        for disk in range(number_disks, 0, -1):
            if contaminado or peg_of_state.get(disk) != peg_of_goal.get(disk):
                misplaced += 1
                contaminado = True
        return misplaced

    metrics = {
        "solution_found": False,
        "nodes_explored": 0,
        "states_visited": 0,
        "nodes_in_frontier": 0,
        "max_depth": 0,
        "cost_total": 0.0,
    }

    root = NodeHanoi(initial_state)
    counter = itertools.count()
    frontier = []
    h0 = _heuristic(root.state)
    heapq.heappush(frontier, (root.path_cost + h0, next(counter), root))
    best_g = {root.state: root.path_cost}
    explored = set()
    solution = None
    max_depth = 0

    while frontier:
        f, _, node = heapq.heappop(frontier)
        if node.state in explored:
            continue
        metrics["nodes_explored"] += 1
        max_depth = max(max_depth, node.depth)
        if _is_goal(node.state):
            solution = node
            metrics["solution_found"] = True
            metrics["cost_total"] = node.path_cost
            break
        explored.add(node.state)
        for child in node.expand(problem):
            if child.state in explored:
                continue
            g = child.path_cost
            if child.state not in best_g or g < best_g[child.state]:
                best_g[child.state] = g
                h = _heuristic(child.state)
                heapq.heappush(frontier, (g + h, next(counter), child))

    metrics["states_visited"] = len(explored)
    metrics["nodes_in_frontier"] = len(frontier)
    metrics["max_depth"] = max_depth
    return solution, metrics


# =====================================================================
# EJECUCIÓN: correr los 7 algoritmos y comparar
# =====================================================================

number_disks = 5  

algo_functions = {
    "BFS": lambda: run_bfs(number_disks),
    "DFS": lambda: run_dfs(number_disks),
    "UCS": lambda: run_ucs(number_disks),
    "DLS": lambda: run_dls(number_disks, limit=31),   # límite = óptimo (2^5 - 1 = 31)
    "IDDFS": lambda: run_iddfs(number_disks),
    "Greedy": lambda: run_greedy(number_disks),
    "A*": lambda: run_astar(number_disks),
}

algorithms = {}   # name -> (solution, metrics)
exec_times = {}   # name -> tiempo real de ejecución (segundos)

for name, fn in algo_functions.items():
    t0 = time.perf_counter()
    sol, met = fn()
    t1 = time.perf_counter()
    algorithms[name] = (sol, met)
    exec_times[name] = t1 - t0

print(f"Resultados para {number_disks} discos (solución óptima teórica = {2**number_disks - 1} movimientos):\n")
for name, (sol, met) in algorithms.items():
    print(f"--- {name}  (tiempo real: {exec_times[name]*1000:.2f} ms) ---")
    for k, v in met.items():
        print(f"  {k}: {v}")
    print()

print("Ranking por tiempo real de ejecución (de más rápido a más lento):")
for name, t in sorted(exec_times.items(), key=lambda kv: kv[1]):
    print(f"  {name}: {t*1000:.2f} ms")
```

```python

# =====================================================================
# ANIMACIÓN: comparar los 7 algoritmos en simultáneo
# =====================================================================

paths = {
    name: ([node.state for node in sol.path()] if sol is not None else [])
    for name, (sol, met) in algorithms.items()
}

peg_x = [1, 2, 3]
colors = plt.cm.tab10.colors
max_height = number_disks + 2

fig, axes = plt.subplots(2, 4, figsize=(14, 7), dpi=80)
axes = axes.flatten()

# Orden de los paneles = mismo orden que el ranking por tiempo real (más rápido primero)
names = [name for name, _ in sorted(exec_times.items(), key=lambda kv: kv[1])]
ranking_pos = {name: i + 1 for i, name in enumerate(names)}  # 1° = más rápido

axes[-1].axis("off")  


def draw_on_ax(ax, state, name, step, total, cost, elapsed, algo_time, rank):
    ax.clear()
    rods = get_rods(state)
    ax.set_xlim(0.3, 3.7)
    ax.set_ylim(0, max_height)
    ax.axis("off")
    ax.set_title(
        f"#{rank}  {name}  (paso {step}/{total}, costo={cost})\n"
        f"t={elapsed*1000:.1f} ms / {algo_time*1000:.1f} ms",
        fontsize=8,
    )
    for x in peg_x:
        ax.plot([x, x], [0, max_height - 0.5], color="black", linewidth=3, zorder=1)
        ax.plot([x - 0.9, x + 0.9], [0, 0], color="black", linewidth=4, zorder=1)
    for peg_idx, rod in enumerate(rods):
        for level, disk in enumerate(rod):
            width = 0.15 + disk * 0.12
            rect = patches.Rectangle(
                (peg_x[peg_idx] - width / 2, level), width, 0.85,
                facecolor=colors[disk % len(colors)], edgecolor="black", zorder=2,
            )
            ax.add_patch(rect)
            ax.text(peg_x[peg_idx], level + 0.42, str(disk),
                     ha="center", va="center", fontsize=8, zorder=3)


# --- Escalado temporal ---
# lo hago agregando un proporcional de tiempo porque sino son mseg y resulta imperceptible la diferencia entre los
# algoritmos de busqueda. 

TOTAL_ANIM_SECONDS = 15.0   # duración total (en pantalla) de la animación completa.
FPS = 15   
total_frames = int(TOTAL_ANIM_SECONDS * FPS)

GAMMA = 0.45  # <1 comprime: acerca los tiempos rápidos y lentos sin invertir el orden. 
def scale_time(t):
    return t ** GAMMA

max_algo_time = max(exec_times.values())
if max_algo_time <= 0:
    max_algo_time = 1e-6  # evitar división por cero si algo fue instantáneo

max_scaled_time = scale_time(max_algo_time)


def update(frame):

    scaled_repr = (frame / (total_frames - 1)) * max_scaled_time

    for ax, name in zip(axes, names):
        path = paths[name]
        algo_time = exec_times[name]
        rank = ranking_pos[name]
        if not path:
            ax.clear()
            ax.axis("off")
            ax.set_title(f"#{rank}  {name}: sin solución\n(t={algo_time*1000:.1f} ms)", fontsize=8)
            continue

        algo_scaled_time = scale_time(algo_time) if algo_time > 0 else 0.0
        progress = min(scaled_repr / algo_scaled_time, 1.0) if algo_scaled_time > 0 else 1.0
        idx = int(progress * (len(path) - 1))
        elapsed = min(progress * algo_time, algo_time)  
        cost = algorithms[name][1]["cost_total"]
        draw_on_ax(ax, path[idx], name, idx, len(path) - 1, cost, elapsed, algo_time, rank)


anim = FuncAnimation(fig, update, frames=total_frames, interval=1000 / FPS, repeat=False)
plt.close(fig)

import matplotlib as mpl
mpl.rcParams['animation.embed_limit'] = 100  

HTML(anim.to_jshtml())
```
